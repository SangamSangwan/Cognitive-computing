```python
# Q1
roll_no = "1024170388"
digits = [int(d) for d in roll_no]
L = [d * 10 for d in digits]
print(L)

L.append(50)
print(L)
print("Number 50 is added in the end")
L.insert(2, 99)
print(L)
print("Number 99 is added at 2nd index")

L.remove(50)
print(L)
L.pop()
print(L)

L.sort()
print(L)
L.sort(reverse=True)
print(L)

print(L[:3])
print(L[-3:])

avg = sum(L) / len(L)
above_avg = [x for x in L if x > avg]
print(above_avg)
```

```python
# Q2
scores = tuple(L[:8])
print(scores)

highest = max(scores)
print(highest, scores.index(highest))
lowest = min(scores)
print(lowest, scores.count(lowest))

reversed_list = list(scores[::-1])
print(reversed_list)
# tuples are immutable, so reverse() can't modify in place; must convert to list

user_score = int(input("Enter a score: "))
if user_score in scores:
    print(scores.index(user_score))
else:
    print("Not present")

try:
    scores[0] = 100
except TypeError as e:
    print(e)
# tuples don't support item assignment because they're immutable, unlike lists

first, second, *rest = scores
print(first, second, rest)
```

```python
# Q3
import random
random.seed(1024170388)

nums = [random.randint(100, 900) for _ in range(100)]

odds = [x for x in nums if x % 2 != 0]
print(len(odds))

evens = [x for x in nums if x % 2 == 0]
print(len(evens))

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

primes = [x for x in nums if is_prime(x)]
print(len(primes))
print(primes)

from collections import Counter
count = Counter(nums)
most_common = count.most_common(1)[0]
print(most_common)
```

```python
# Q4
digits8 = digits[:8]
A = {d * 7 for d in digits8}
B = {d * 9 for d in digits8}
print(A)
print(B)

print(A | B)
print(A & B)

print(A - B)
print(B - A)
# difference() gives elements only in one set; symmetric_difference() gives elements in either but not both

print(A ^ B)

print(A.issubset(B))
print(B.issuperset(A))

X = int(input("Enter value to discard: "))
A.discard(X)
print(A)
# discard() doesn't raise an error if the value isn't present, unlike remove()
```

```python
# Q5
my_dict = {
    "name": "Sangam Sangwan",
    "roll_no": "1024170388",
    "branch": "Computer Science Engineering",
    "age": 20,
    "city": "Charkhi Dadri"
}

my_dict["location"] = my_dict.pop("city")
print(my_dict)

my_dict["cgpa"] = 8.8
print(my_dict)

my_dict["age"] += 1
print(my_dict)

dict_pop = my_dict.copy()
popped_value = dict_pop.pop("branch")
print(dict_pop, popped_value)

dict_del = my_dict.copy()
del dict_del["branch"]
print(dict_del)
# pop() returns the removed value, del returns nothing

for k, v in my_dict.items():
    print(f"{k} \u2192 {v}")

if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email not found")

friend_dict = {
    "name": "Samyak Jain",
    "roll_no": "1024170400",
    "branch": "Computer Science Engineering",
    "age": 20,
    "location": "Delhi"
}

merged = {**my_dict, **friend_dict}
print(merged)
# when keys overlap, values from the second dictionary (friend_dict) win

string_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print(string_dict)
```
