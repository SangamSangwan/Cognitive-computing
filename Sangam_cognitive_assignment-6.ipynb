## Q1

```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
```

**Q1(a)**

```python
print(temperature + 2)
```

**Q1(b)**

```python
print((9/5) * temperature + 32)
```

**Q1(c)**

```python
print(temperature[temperature > 32])
```

**Q1(d)**

```python
print(np.sum(temperature > 32))
```

**Q1(e)**

Vectorization applies an operation to the whole array at once in optimized C code, instead of Python handling one element per loop iteration, so it is faster and needs less code. Boolean indexing selects or counts elements with a single mask (`temperature > 32`) instead of a loop with `if` conditions. Together they make processing shorter, cleaner and much quicker, especially on large data.

## Q4

```python
X = np.array([[6, 70, 3], [5, 50, 6], [8, 80, 2], [4, 30, 8]])
y = np.array([40, 65, 30, 85])
```

**Q4(a)**

```python
print(X.shape, X.ndim)
```

**Q4(b)**

```python
print(X.T)
```

The transpose swaps rows and columns: rows become features and columns become users (3 × 4 instead of 4 × 3).

**Q4(c)**

```python
XtX = X.T @ X
print(XtX)
```

**Q4(d)**

```python
XtX_inv = np.linalg.inv(XtX)
print(XtX_inv)
```

**Q4(e)**

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print(beta)
```

**Q4(f)**

Each coefficient is the change in predicted assistance score for a one-unit increase in that feature (sleep hours, activity level, stress level), keeping the other two fixed. A negative value lowers the score, a positive value raises it.

**Q4(g)**

```python
new_user = np.array([5, 40, 7])
print(new_user @ beta)
```
