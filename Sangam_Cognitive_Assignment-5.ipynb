## Q2. Basic NumPy array questions

### a) Reverse the array

```python
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("Reversed array:", reversed_arr)
```

### b) Most frequent value and its indices

```python
def most_frequent(arr):
    values, counts = np.unique(arr, return_counts=True)
    max_count = counts.max()
    freq_val = values[counts == max_count][0]   # most frequent value (first if tie)
    indices = np.where(arr == freq_val)[0]
    return freq_val, indices

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

val_x, idx_x = most_frequent(x)
print(f"x: most frequent value = {val_x}, indices = {idx_x}")

val_y, idx_y = most_frequent(y)
print(f"y: most frequent value = {val_y}, indices = {idx_y}")
```

## Q4. 1-D array with linspace

```python
Sangam = np.linspace(10, 100, 25)
print("Array:", Sangam)

print("Dimensions (ndim):", Sangam.ndim)
print("Shape:", Sangam.shape)
print("Total elements (size):", Sangam.size)
print("Data type:", Sangam.dtype)
print("Total bytes (nbytes):", Sangam.nbytes)

# Transpose using reshape() 
transpose_reshape = Sangam.reshape(25, 1)
print("Transpose using reshape():\n", transpose_reshape)

# Transpose using .T attribute
transpose_T = Sangam.T
print("Transpose using .T attribute:", transpose_T)
print("Shape after .T:", transpose_T.shape)

# .T has no effect on a 1-D array (shape stays (25,));
# reshape(25,1) genuinely produces a 2-D column vector.
# So .T alone cannot replicate the reshape() result for a 1-D array.
```

## Q5. 2-D array — stats, reshape, resize

```python
ucs420_Sangam = np.array([[10, 20, 30, 40],
                           [50, 60, 70, 80],
                           [90, 15, 20, 35]])
print("Array:\n", ucs420_Sangam)

print("Mean:", np.mean(ucs420_Sangam))
print("Median:", np.median(ucs420_Sangam))
print("Max:", np.max(ucs420_Sangam))
print("Min:", np.min(ucs420_Sangam))
print("Unique elements:", np.unique(ucs420_Sangam))

# Reshape to 4 rows x 3 columns
reshaped_ucs420_Sangam = ucs420_Sangam.reshape(4, 3)
print("\nReshaped (4x3):\n", reshaped_ucs420_Sangam)

# Resize to 2 rows x 3 columns (drops/truncates extra elements)
resized_ucs420_Sangam = np.resize(ucs420_Sangam, (2, 3))
print("\nResized (2x3):\n", resized_ucs420_Sangam)
```
